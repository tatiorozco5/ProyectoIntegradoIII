```python
import pandas as pd
import numpy as np
```

#Carga del dataset desde la URL

```python
url = "https://www.datos.gov.co/api/views/shu7-uqse/rows.csv?accessType=DOWNLOAD"

df = pd.read_csv(url)

# Visualizar primeras filas
df.head()
```

```python
# Tamaño del dataset
df.shape
```

```python
# Información general
df.info()
```

```python
# Estadísticas descriptivas
df.describe(include='all')
```

#Renombrar columnas

Se renombran las columnas para facilitar el trabajo en Python y Power BI.

```python
df.columns = [
    'fecha_inicio',
    'fecha_final',
    'departamento',
    'municipio',
    'fuerza_entidad',
    'entidad_vinculada',
    'proceso_gestion_riesgo',
    'actividad_realizada',
    'medios_equipos_empleados',
    'cantidad_medios_empleados',
    'personal_empleado'
]

df.columns
```

#Verificación de valores nulos

```python
df.isnull().sum()
```

#Tratamiento de valores nulos

##Variables categóricas

Se reemplazan valores nulos por “No especificado”.

```python
columnas_texto = [
    'entidad_vinculada',
    'medios_equipos_empleados'
]

for col in columnas_texto:
    df[col] = df[col].fillna('No especificado')
```

##Variables numéricas

Se reemplazan valores nulos por 0.

```python
df['cantidad_medios_empleados'] = df['cantidad_medios_empleados'].fillna(0)
```

#Conversión de tipos de datos

```python
df['fecha_inicio'] = pd.to_datetime(df['fecha_inicio'], errors='coerce')
df['fecha_final'] = pd.to_datetime(df['fecha_final'], errors='coerce')
```

##Conversión de variables numéricas

```python
df['cantidad_medios_empleados'] = df['cantidad_medios_empleados'].astype(int)
df['personal_empleado'] = df['personal_empleado'].astype(int)
```

#Verificación de duplicados

```python
df.duplicated().sum()
```

#Eliminación de duplicados

```python
df = df.drop_duplicates()
```

#Estandarización de texto

Se normalizan textos para evitar inconsistencias.

```python
columnas_categoricas = [
    'departamento',
    'municipio',
    'fuerza_entidad',
    'entidad_vinculada',
    'proceso_gestion_riesgo',
    'actividad_realizada'
]

for col in columnas_categoricas:
    df[col] = df[col].str.strip().str.upper()
```

#Validación final

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.head()
```

#Creación de variables temporales

```python
df['anio'] = df['fecha_inicio'].dt.year
df['mes'] = df['fecha_inicio'].dt.month
df['nombre_mes'] = df['fecha_inicio'].dt.month_name()
```

#Exportar dataset limpio

```python
df.to_csv('dataset_limpio.csv', index=False)
```
